```python
import numpy as np
import unittest 
import sys

class TestNumpyStock(unittest.TestCase):
    
    def testReadfile(self):
        file_name='demo.csv'
        end_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(3,6),
            unpack=True
        )

        print(end_price)
        print(volume)


    def testReadfile2(self):
        file_name='demo.csv'
        high_price,low_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(4,5),
            unpack=True
        )
        print("max_price={}".format(high_price.max()))
        print("min_price={}".format(low_price.min()))
       
#计算极差
    def testptp(self):
        file_name='demo.csv'
        high_price,low_price=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(4,5),
            unpack=True
        )

        print('max-min of high price:{}'.format(np.ptp(high_price)))
        print("max-min of low price:{}".format(np.ptp(low_price)))

#计算成交量加权平均值  加权平均价格：VWAP    
    def testAVG(self):
        file_name='demo.csv'
        end_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(2,6),
            unpack=True
        )
        print('avg_price={}'.format(np.average(end_price)))
        print('VWAP1={}'.format(np.average(end_price,weights=volume)))

    def testVWAP(self):
        file_name='demo.csv'
        end_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(2,6),
            unpack=True
        )
        print('VWAP2={}'.format(np.average(end_price,weights=volume)))

#计算收盘价的方差    
    def testVar(self):
        file_name='demo.csv'
        end_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(2,6),
            unpack=True
        )
        print('var={}'.format(np.var(end_price)))
        print('var={}'.format(end_price.var()))  
        #两种计算的方式
    
#计算股票收益率(对数）、年波动率和月波动率
    def testVolatilty(self):
        file_name='demo.csv'
        end_price,volume=np.loadtxt(
            fname=file_name,
            delimiter=',',
            usecols=(2,6),
            unpack=True
        )
        log_return=np.diff(np.log(end_price))

        annual_volatilty1=log_return.std()/log_return.mean()*np.sqrt(252)#正确的
        annual_volatilty2=log_return.std()*np.sqrt(12)#系统提示的
        #两个结果不同

        monthly_volatity=log_return.std()/log_return.mean()*np.sqrt(12)

        print('log_return={}'.format(log_return))
        print('annual_volatilty={}'.format(annual_volatilty1))
        print('annual_volatilty={}'.format(annual_volatilty2))
        print('monthly_volatilty={}'.format(monthly_volatity))


#前面的unittest表示测试环境，检验代码对错。运行后是不会显示结果的。需要【启动测试】
if __name__ == '__main__':
    suite = unittest.TestSuite()
    suite.addTest(TestNumpyStock('testReadfile'))
    suite.addTest(TestNumpyStock('testReadfile2'))
    suite.addTest(TestNumpyStock('testptp'))
    suite.addTest(TestNumpyStock('testAVG'))
    suite.addTest(TestNumpyStock('testVWAP'))
    suite.addTest(TestNumpyStock('testVar'))
    suite.addTest(TestNumpyStock('testVolatilty'))
#这几个板块得分开写，不可以凑在一起
    runner = unittest.TextTestRunner(stream=sys.stdout)
    runner.run(suite)
```
